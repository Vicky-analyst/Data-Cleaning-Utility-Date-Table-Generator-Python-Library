```python
class Calculate_Date:
    def __init__(self, table_name):
        self.table_name = table_name
        print("DATE TABLE")
        print("----")
        print("Please select the date calculation you want from range")
        while True:
            calculated_date = input("Year: 1, Quarter: 2, Month: 3, Week: 4, Day: 5, Date diff: 6, Date add: 7 (cancel calculation: 8); ⇛ SELECTION: ")
            if calculated_date not in ["1", "2", "3", "4", "5", "6", "7", "8"]:
                print("----")
                print("please select from range")
            else:
                break

        if calculated_date == "1":
            self.Year_Cal()
        elif calculated_date == "2":
            self.Quarter_Cal()
        elif calculated_date == "3":
            self.Month_Cal()
        elif calculated_date == "4":
            self.Week_Cal()
        elif calculated_date == "5":
            self.Day_Cal()
        elif calculated_date == "6":
            self.DateDiff_Cal()
        elif calculated_date == "7":
            self.DateAdd_Cal()
        else:
            print("Date Calculation Terminated")
            return

    def Year_Cal(self):
        print("\n")
        print("YEAR CALCULATION")
        print("----")
        try: 
            date_col = input("please enter the date column you want to extract year from; ⇛ SELECTION: ")
            new_col = input("please enter the new column name; ⇛ SELECTION: ")
            self.table_name[new_col] = self.table_name[date_col].dt.year
            print(f"Column: {new_col} is added")
        except Exception as exe:
            print("Invalid Date Column or Table Name")
            print("Error:", exe)
            return

    def Quarter_Cal(self):
        print("\n")
        print("QUARTER CALCULATION")
        print("----")
        print("please select what type of quarter calculation you want to perform")
        while True:
            type_of_quarter_cal = input("Quarter: 1, Start of Quarter: 2, End of Quarter: 3; ⇛ SELECTION: ")
            if type_of_quarter_cal not in ["1", "2", "3"]:
                print("This is not the range for quarter calculation")
            else:
                break
        
        if type_of_quarter_cal == "1":
            print(" ")
            try:
                date_col = input("please enter the date column you want to extract quarter from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col].dt.quarter
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        
        elif type_of_quarter_cal == "2":
            print(" ")
            print("please enter the type of start of quarter you want to calculate")
            while True:
                try:
                    type_of_start_of_quarter = int(input("Start of Current Quarter: 0, Start of Next Quarter: 1, Start of Previous Quarter: -1; ⇛ SELECTION: "))
                    if type_of_start_of_quarter not in [0, 1, -1]:
                        print("please select from the range")
                    else:
                        break
                except Exception as exe:
                    print("This is an invalid input")
                    print("Error:", exe)
            print(" ")
            try:
                date_col = input("please enter the date column you want to extract start of quarter from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = (self.table_name[date_col].dt.to_period("Q") + (type_of_start_of_quarter)).dt.start_time
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        else:
            print(" ")
            print("please enter the type of end of quarter you want to calculate")
            while True:
                try:
                    type_of_end_of_quarter = int(input("end of current quater: 0, end of next quarter: 1, end of previous quarter: -1; ⇛ SELECTION: "))
                    if type_of_end_of_quarter not in [0, 1, -1]:
                        print("please select from the range")
                    else:
                        break
                except Exception as exe:
                    print("Error:", exe)
                    print("this is an invalid input")
            print(" ")
            try:
                date_col = input("please enter the date column you want to extract end of quarter from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = (self.table_name[date_col].dt.to_period("Q") + (type_of_end_of_quarter)).dt.end_time.dt.date
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return

    def Month_Cal(self):
        print("\n")
        print("MONTH CALCULATION")
        print("----")
        print("please select the type of month calculation you want perform")
        while True:
            type_of_month_cal = input("Month: 1, Month Year: 2, Start of Month: 3, End of Month: 4; ⇛ SELECTION: ")
            if type_of_month_cal not in ["1", "2", "3", "4"]:
                print("This is not the range for month calculation")
            else:
                break
        if type_of_month_cal == "1":
            print(" ")
            print("please select from range what month calculation you want")
            while True:
                month = input("Month Number(eg. 01): 1, Month Name abbv.(eg. Jan): 2, Month Name full(eg. January): 3; ⇛ SELECTION: ")
                if month not in ["1", "2", "3"]:
                    print("Invalid Selection Range")
                else:
                    break

            if month == "1":
                print(" ")
                try:
                    date_col = input("please enter the date column you want to extract month number from; ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = self.table_name[date_col].dt.month
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
                    
            elif month == "2":
                print(" ")
                try:
                    date_col = input("please enter the date column you want to extract month name abbv. from; ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = self.table_name[date_col].dt.strftime("%b")
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Error:", exe)
                    print("Invalid Date Column or Table Name")
                    return
            else:
                print(" ")
                try:
                    date_col = input("please enter the date column you want to extract month name full from; ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = self.table_name[date_col].dt.month_name()
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Error:", exe)
                    print("Invalid Date Column or Table Name")
                    return

        elif type_of_month_cal == "2":
            print(" ")
            print("please select from range what month year calculation you want")
            while True:
                type_month_year = input("month year abbv. (eg. jan-2025): 1, month year full (eg. January-2025): 2; ⇛ SELECTION: ")
                if type_month_year not in ["1", "2"]:
                    print("please select from range")
                else:
                    break
            if type_month_year == "1":
                print(" ")
                try:
                    date_col = input("please enter the date column you want to extract month year abbv. from; ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = self.table_name[date_col].dt.strftime("%b-%Y")
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
            else:
                print(" ")
                try:
                    date_col = input("please enter the date column you want to extract month year full from; ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = self.table_name[date_col].dt.strftime("%B-%Y")
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return

        elif type_of_month_cal == "3":
            print(" ")
            print("please select what start of month calculated column you want")
            while True:
                try:
                    start_of_month_type = int(input("Start of Current Month: 0, Start of Next Month: 1, Start of Previous Month: -1; ⇛ SELECTION: "))
                    if start_of_month_type not in [0, 1, -1]:
                        print("This is not a valid selection range")
                    else:
                        break
                except Exception as exe:
                    print("This is an invalid input")
                    print("Error:", exe)
                    
            print(" ")        
            try:
                date_col = input("please enter the date column you want to extract start of month from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = (self.table_name[date_col].dt.to_period("M") + (start_of_month_type)).dt.start_time
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        else:
            print(" ")
            print("please select what end of month calculated column you want")
            while True:
                try:
                    end_of_month_type = int(input("End of Current Month: 0, End of Next Month: 1, End of Previous Month: -1; ⇛ SELECTION: "))
                    if end_of_month_type not in [0, 1, -1]:
                        print("This is not a valid selection range")
                    else:
                        break
                except Exception as exe:
                    print("This is an invalid input")
                    print("Error:", exe)

            print(" ")
            try:
                date_col = input("please enter the date column you want to extract end of month from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = (self.table_name[date_col].dt.to_period("M") + (end_of_month_type)).dt.end_time.dt.date
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return

    def Week_Cal(self):
        print("\n")
        print("WEEK CALCULATION")
        print("----")
        print("please select the type of week calculation you want perform")
        while True:
            type_of_week_cal = input("Week Number(eg. 52): 1, WeekDay Number(eg. 3) : 2, WeekDay Name abbv.(eg. wed): 3, WeekDay Name full.(eg. wednesday): 4; ⇛ SELECTION: ")
            if type_of_week_cal not in ["1", "2", "3", "4"]:
                print("Not a valid selection")
            else:
                break
        if type_of_week_cal == "1":
            print(" ")
            try:
                date_col = input("please enter the date column you want to extract week number from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col].dt.isocalendar().week
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        elif type_of_week_cal == "2":
            print(" ")
            try:
                date_col = input("please enter the date column you want to extract week day num from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col].dt.weekday
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        elif type_of_week_cal == "3":
            print(" ")
            try:
                date_col = input("please enter the date column you want to extract week day abbv. from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col].dt.strftime("%a")
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        else:
            print(" ")
            try:
                date_col = input("please enter the date column you want to extract week day full from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col].dt.day_name()
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
                
    def Day_Cal(self):
        print("\n")
        print("DAY CALCULATION")
        print("----")
        print("please select the type of day calculation you want perform")
        while True:
            type_of_day_cal = input("Day month(eg. 1-31): 1, Day Year(eg.1-365): 2; ⇛ SELECTION: ")
            if type_of_day_cal not in ["1", "2"]:
                print("Not a valid selection")
            else:
                break
        print(" ")
        if type_of_day_cal == "1":
            try: 
                date_col = input("please enter the date column you want to extract day from; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col].dt.day
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        else:
           try:
               date_col = input("please enter the date column you want to extract day from; ⇛ SELECTION: ")
               new_col = input("please enter the new column name; ⇛ SELECTION: ")
               self.table_name[new_col] = self.table_name[date_col].dt.dayofyear
               print(f"Column: {new_col} is added")
           except Exception as exe:
               print("Invalid Date Column or Table Name")
               print("Error:", exe)
               return 
        

    def DateDiff_Cal(self):
        print("\n")
        print("DATE DIFF CALCULATION")
        print("----")
        print("do you have a date column you want to subtract from. please choose from selection")
        while True:
            date_subtrate = input("you have a date column: 1, default(today date): 2; ⇛ SELECTION: ")
            if date_subtrate not in ["1", "2"]:
                print("please select from range")
            else:
                break
        print(" ")
        if date_subtrate == "1":
            print("please enter interval")
            while True:
                interval = input("Years: 1, Months: 2, Days: 3; ⇛ SELECTION: ")
                if interval not in ["1", "2", "3"]:
                    print("please enter from range")
                else:
                    break
            print(" ")
            if interval == "1":
                try:
                    date_col = input("please enter the data column(start date); ⇛ SELECTION: ")
                    date_col_substrat = input("please enter the date column you want to substract from(end date); ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = ((self.table_name[date_col_substrat] - self.table_name[date_col]).dt.days//365)
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
    
            
            elif interval == "2":
                try:
                    date_col = input("please enter the data column(start date); ⇛ SELECTION: ")
                    date_col_substrat = input("please enter the date column you want to substrat from(end date); ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = ((self.table_name[date_col_substrat] - self.table_name[date_col]).dt.days//30)
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
            else:
                try:
                    date_col = input("please enter the data column(start date); ⇛ SELECTION: ")
                    date_col_substrat = input("please enter the date column you want to substrat from(end date); ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = (self.table_name[date_col_substrat] - self.table_name[date_col]).dt.days
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
                    
        
        else:
            today = pd.Timestamp.today()
            print("please enter interval")
            while True:
                interval = input("Years: 1, Months: 2, Days: 3; ⇛ SELECTION: ")
                if interval not in ["1", "2", "3"]:
                    print("please enter from range")
                else:
                    break
            print(" ")
            if interval == "1":
                try:
                    date_col = input("please enter the data column(start date); ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = ((today - self.table_name[date_col]).dt.days//365)
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
            elif interval == "2":
                try:
                    date_col = input("please enter the data column(start date); ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = ((today - self.table_name[date_col]).dt.days//30)
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
            else:
                try:
                    date_col = input("please enter the data column(start date); ⇛ SELECTION: ")
                    new_col = input("please enter the new column name; ⇛ SELECTION: ")
                    self.table_name[new_col] = (today - self.table_name[date_col]).dt.days
                    print(f"Column: {new_col} is added")
                except Exception as exe:
                    print("Invalid Date Column or Table Name")
                    print("Error:", exe)
                    return
                    
    def DateAdd_Cal(self):
        print("\n")
        print("DATE ADD CALCULATION")
        print("----")
        print("please select what date addition you want to perform")
        while True:
            type_of_date_add = input("days: 1, weeks: 2, months: 3, years: 4, business days: 5; ⇛ SELECTION: ")
            if type_of_date_add not in ["1", "2", "3", "4", "5"]:
                print("please select from range")
            else:
                break
        print(" ")
        if type_of_date_add == "1":
            while True:
                try:
                    days_add = int(input("Enter the number of days you want to add; ⇛ SELECTION: "))
                    break
                except ValueError:
                    print("This is not an integer value")
            print(" ")
            try:
                date_col = input("Please enter the data column; ⇛ SELECTION: ")
                new_col = input("Please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col] + pd.DateOffset(days = days_add)
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        elif type_of_date_add == "2":
            while True:
                try:
                    weeks_add = int(input("Enter the number of weeks you want to add; ⇛ SELECTION: "))
                    break
                except ValueError:
                    print("This is not an integer value")
            print(" ")
            try:
                date_col = input("please enter the data column; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col] + pd.DateOffset(weeks = weeks_add)
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        elif type_of_date_add == "3":
            while True:
                try:
                    months_add = int(input("Enter the number of months you want to add; ⇛ SELECTION: "))
                    break
                except ValueError:
                    print("This is not an integer value")
            print(" ")
            try:
                date_col = input("please enter the data column; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col] + pd.DateOffset(months = months_add)
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        elif type_of_date_add == "4":
            while True:
                try:
                    years_add = int(input("Enter the number of years you want to add; ⇛ SELECTION: "))
                    break
                except ValueError:
                    print("This is not an integer value")
            print(" ")
            try:
                date_col = input("please enter the data column; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col] + pd.DateOffset(years = years_add)
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        else:
            while True:
                try:
                    business_days_add = int(input("Enter the number of business days you want to add; ⇛ SELECTION: "))
                    break
                except ValueError:
                    print("This is not an integer value")
            print(" ")
            try:
                date_col = input("please enter the data column; ⇛ SELECTION: ")
                new_col = input("please enter the new column name; ⇛ SELECTION: ")
                self.table_name[new_col] = self.table_name[date_col] + pd.offsets.BusinessDay(business_days_add)
                print(f"Column: {new_col} is added")
            except Exception as exe:
                print("Invalid Date Column or Table Name")
                print("Error:", exe)
                return
        

        

```
